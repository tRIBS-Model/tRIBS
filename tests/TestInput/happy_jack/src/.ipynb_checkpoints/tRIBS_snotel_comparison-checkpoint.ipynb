# Comparison of reference Happy Jack results to Snotel Data

__Import relevant packages including [tP4](https://github.com/WrenRaming/tP4) python package for visualizing and comparing model results__

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
from tribsmodel import Model as model
```

```python
m = model()
# change to main directory
os.chdir('..')
```

```python
m.read_input_file('src/template.in')
m.outfilename['value']='results/reference/HJ_results'
```

__Because this is single point model the best approach to visualizing results is to simply look a the .pixel files which represent the single active computaional element__

```python
m.Results.get_element_results()# read in the .pixel file
m.Results.get_element_water_balance('water_year') # calculate waterbalance
```

```python
pixel = m.Results.element['0'] #returns pandas data frame from dict specified by node ID, this case its 0
waterbalance = pixel['waterbalance']
pixel = pixel['pixel']
```

__Read in the snotel data__

```python
# Specify the path to snotel data
file_path = 'data/Snotel/bcqc_34.75000_-111.41000.txt'

# precip is in inches, T is temperature in F, and SWE is snow water equivalent in inches
column_names = ['year', 'month', 'day', 'precip', 'maxT','minT','meanT','swe']
snotel = df = pd.read_csv(file_path,  sep='\s+', header=None, na_values=['nan'], names=column_names)

#convert year, month, day to datetime and drop columns
snotel['date'] = pd.to_datetime(snotel [['year', 'month', 'day']])
snotel  = snotel .drop(['year', 'month', 'day'], axis=1)

#convert swe to cm 
snotel['swe_cm'] = snotel['swe'] * 2.54
```

__Compare time series__

```python
fig,ax = plt.subplots()
plt.plot(snotel.date,snotel.swe_cm,label = 'Snotel')
plt.plot(pixel.Time,pixel.SnWE_cm,label = 'tRIBS')
plt.xlabel('Date')
plt.ylabel('Snow Water Equivalent (cm)')
plt.legend(loc='upper left')
```

__We can also examine the water balance of the refrence case as follows:__

```python
m.Results.plot_water_balance(waterbalance)
```

__The number above bar represent the diffrence in perfect water balance closure and the model results.__
